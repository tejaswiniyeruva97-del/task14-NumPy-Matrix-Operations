```python
import numpy as np

# Creating Matrix A
A = np.array([[1, 2],
              [3, 4]])

# Creating Matrix B
B = np.array([[5, 6],
              [7, 8]])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B

print("Matrix Addition:")
print(addition)
```

```python
subtraction = A - B

print("Matrix Subtraction:")
print(subtraction)
```

```python
multiplication = np.dot(A, B)

print("Matrix Multiplication:")
print(multiplication)
```

```python

transpose_A = A.T
transpose_B = B.T

print("Transpose of Matrix A:")
print(transpose_A)

print("\nTranspose of Matrix B:")
print(transpose_B)
```

```python

import numpy as np

# Create two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Display matrices
print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Addition
print("\nMatrix Addition:")
print(A + B)

# Subtraction
print("\nMatrix Subtraction:")
print(A - B)

# Matrix Multiplication
print("\nMatrix Multiplication:")
print(np.dot(A, B))

# Transpose
print("\nTranspose of Matrix A:")
print(A.T)

print("\nTranspose of Matrix B:")
print(B.T)
```
